```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Dados
data = {
    'Nome': ['Alice', 'Joao', 'Charlie', 'David', 'Eva', 'Diego', 'Denize', 'Claudio'],
    'Idade': [25, 30, 35, 40, 45, 60, 22, 24],
    'Profissão': ['Engenheiro', 'Médico', 'Professor', 'Advogado', 'Médico','Engenheiro', 'Estudante','Estudante'],
    'Salário': ['4500', '8000', '5000', '10000', '12000','15000', '1200','1500'],
    'Limite_Credito': ['2500', '4000', '4000', '1000', '10000','2000', '500','250'],
    'Historico_Inadimplencia': ['0', '0', '0', '1', '0','1', '0','1'],
    'Estado_Civil': ['Casamento', 'Casamento', 'Solteiro', 'Solteiro', 'Casamento','Solteiro', 'Solteiro','Solteiro'],
    'Imovel_Proprio': ['0', '0', '0', '1', '1','1', '0','0']
}

df = pd.DataFrame(data)

num_cols = ['Salário', 'Limite_Credito', 'Historico_Inadimplencia', 'Imovel_Proprio']
df[num_cols] = df[num_cols].astype(int)

# Salário vs Limite de Crédito
sns.scatterplot(data=df, x='Salário', y='Limite_Credito', hue='Profissão')
plt.title('Salário vs Limite de Crédito')
plt.show()

# Limite de Crédito por Profissão
sns.boxplot(data=df, x='Profissão', y='Limite_Credito')
plt.title('Distribuição do Limite de Crédito por Profissão')
plt.xticks(rotation=45)
plt.show()

# Inadimplência vs Limite de Crédito
sns.boxplot(data=df, x='Historico_Inadimplencia', y='Limite_Credito')
plt.title('Limite de Crédito vs Histórico de Inadimplência')
plt.xticks([0, 1], ['Sem Inadimplência', 'Com Inadimplência'])
plt.show()
```

```python
O primeiro gráfico utilizado foi o de dispersão com cores por profissão, existe uma relação clara entre o salário e o limite de crédito.
Profissionais com salários mais altos tendem a receber limites maiores, profissões como médico e engenheiro se destacam nesse grupo.
```

```python
Depois utilizei o Boxplot por profissão.
A profissão influencia diretamente o limite de crédito, profissionais com cargos mais valorizados no mercado, como médicos, apresentam limites mais altos. 
Já estudantes, por exemplo, têm limites significativamente mais baixos.
```

```python
Por último usei novamente o boxplot de inadimplência
Clientes com histórico de inadimplência possuem limites de crédito mais baixos, indicando que esse fator impacta negativamente a liberação de crédito.
```

```python
Com base na análise dos gráficos, podemos concluir que os principais fatores que impulsionam o limite de crédito são: o Salário, profissão e histórico
de inadimplência que afeta negativamente o limite disponível.
Esses fatores devem ser levados em consideração por instituições financeiras na hora de definir limites de crédito personalizados.
```
